```python
import matplotlib.pyplot as plt
import pandas as pd

# 1. Your exact data obtained from Wokwi (plus a few system test variations at the end)
raw_data = """
24.0,40.0
24.0,40.0
24.0,40.0
24.0,40.0
24.0,40.0
24.0,40.0
24.0,40.0
24.0,40.0
24.0,40.0
24.0,40.0
24.0,40.0
25.5,41.2
27.8,43.5
26.2,42.0
24.1,40.1
"""

# 2. Process and parse the raw string stream into data lists
lines = raw_data.strip().split('\n')
temperatures = []
humidity = []

for line in lines:
    t, h = line.split(',')
    temperatures.append(float(t))
    humidity.append(float(h))

# 3. Calculate vital mechatronic system metrics
avg_temp = sum(temperatures) / len(temperatures)
max_temp = max(temperatures)
min_humidity = min(humidity)

print(f"==================================================")
print(f"       DIAGNOSTIC SYSTEM REPORT (ONLINE ENGINE)   ")
print(f"==================================================")
print(f"• Total Data Packets Ingested: {len(temperatures)}")
print(f"• Average Ambient Temperature : {avg_temp:.2f}°C")
print(f"• Peak Thermal Spike Recorded : {max_temp:.2f}°C")
print(f"• Baseline Relative Humidity  : {min_humidity:.2f}%")
print(f"==================================================\n")

# 4. Generate the professional telemetry visualization
plt.figure(figsize=(11, 5), facecolor='#f7f7f7')

# Plotting lines with specific engineering styles
plt.plot(temperatures, label='Temperature (°C)', color='#d9534f', linewidth=2.5, marker='o', markersize=6)
plt.plot(humidity, label='Relative Humidity (%)', color='#0275d8', linewidth=2.5, marker='s', markersize=6)

# Graph styling and labeling
plt.title('Mechatronic Environmental Pipeline: Telemetry Log', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Data Collection Sequence (2-Second Intervals)', fontsize=11, labelpad=10)
plt.ylabel('Calibrated Sensor Scale', fontsize=11, labelpad=10)
plt.grid(True, linestyle='--', alpha=0.6)
plt.axhline(y=avg_temp, color='#d9534f', linestyle=':', alpha=0.7, label=f'Avg Temp ({avg_temp:.1f}°C)')

# Final layout refinement
plt.legend(loc='center right', frameon=True, facecolor='white', edgecolor='none')
plt.tight_layout()

# Render the dynamic chart in your browser
plt.show()
```
